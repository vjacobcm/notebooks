```python
import torch 
```

```python
names = open('names.txt', 'r').read().splitlines()
```

```python
names[:100]
```

```python
char_lookup = set(''.join(names))
char_lookup
char_lookup_list = ['.'] + sorted(list(char_lookup))
lookup_dict = {s:i  for i,s in enumerate(char_lookup_list)}
lookup_dict

reverse_lookup_dict = {s:i for s,i in enumerate(char_lookup_list)}

reverse_lookup_dict
```

```python
N = torch.zeros((27,27), dtype = torch.int32)
N
```

```python
for name in names:
    chars = ['.'] + list(name) + ['.']
    for ch1, ch2 in zip(chars,chars[1:]):
        bigram = (lookup_dict[ch1] ,lookup_dict[ch2])
        N[bigram]+=1
```

```python
N
```

```python

g = torch.Generator().manual_seed(2147483647)
p = torch.rand(3,generator = g)
p = p/p.sum()
p
```

```python
print(reverse_lookup_dict[torch.multinomial (N[0]/N[0].sum(), num_samples = 1, replacement=True, generator=g).item()])
```

```python
P = N.float()
P = P / P.sum(1,keepdim = True)
```

```python
idx = 0
name = ''
while True:
    current_row = P[idx]
    new_char = reverse_lookup_dict[torch.multinomial (P[idx], num_samples = 1, replacement=True, generator=g).item()]
    if new_char == '.':
        break
    name += new_char
    idx = lookup_dict[new_char]
name
```

```python
import torch.nn.functional as F
xs, ys = [],[]

for name in names:
    chars = ['.'] + list(name) + ['.']
    for char1, char2 in zip(chars, chars[1:]):
        xs.append(lookup_dict[char1])
        ys.append(lookup_dict[char2])

xs = torch.tensor(xs)
ys = torch.tensor(ys)

print(xs.size())
print(ys.size())

W = torch.randn((27,27),generator=g,requires_grad=True)

input_idxs = torch.arange(xs.nelement())
F.one_hot(xs[1], num_classes=27).float()
```

```python
# GRADIENT DESCENT
for n in range(10):
    xs_encoded = F.one_hot(xs, num_classes=27).float()
    counts = (xs_encoded @ W).exp() #SOFTMAX
    probabilities = counts/counts.sum(1, keepdims = True) #SOFTMAX
    loss = - probabilities[input_idxs, ys].log().mean()
    print(loss.item())

    W.gradient = None
    loss.backward()

    W.data += -25 * W.grad

    
```

```python
for i in range(10):
    out = []
    idx = 0
    while True:
        xenc = F.one_hot(torch.tensor([idx]),num_classes=27).float()
        logits = xenc@W
        counts = logits.exp()
        p = counts/counts.sum(1,keepdims=True)

        idx = torch.multinomial(p, num_samples=1, replacement=True, generator=g).item()

        out.append(reverse_lookup_dict[idx])
        if  reverse_lookup_dict[idx] == '.':
            break
    print(''.join(out))
```
